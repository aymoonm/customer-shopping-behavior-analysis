```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.mean()))
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns = { 'purchase_amount_(usd)' : 'purchase_amount'})
df.columns
```

```python
# Create a column age_group
labels = ['Young Adult', 'Adult','Middle Age','Senior']
df['age_group']  = pd.qcut(df['age'], q = 4, labels = labels)
```

```python
df[['age','age_group']].head(10)    
```

```python
# Create column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Weekly': 7,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months' : 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
import pandas as pd
from sqlalchemy import create_engine
import urllib

# 1. إعداد بيانات الاتصال بـ SQL Server الخاص بكِ
SERVER_NAME = r'MO_AYMAN\SQLEXPRESS'  
DATABASE_NAME = 'customer_behavior'

# نص الاتصال (Connection String) باستخدام Windows Authentication
connection_string = f"DRIVER={{ODBC Driver 17 for SQL Server}};SERVER={SERVER_NAME};DATABASE={DATABASE_NAME};Trusted_Connection=yes;"

try:
    # تحويل نص الاتصال لصيغة تفهمها مكتبة SQLAlchemy
    params = urllib.parse.quote_plus(connection_string)
    engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")
    print("تم إعداد محرك الاتصال (Engine) بنجاح.")
    
    # 2. قراءة ملف البيانات الخاص بكِ من المسار المحدد
    file_path = r'C:\Protfolio_Project\Project_3\customer_shopping_behavior.csv'
    df = pd.read_csv(file_path)
    print(f"تم قراءة ملف البيانات بنجاح. عدد الصفوف: {len(df)}")

    # 3. رفع البيانات إلى SQL Server
    # سيتم إنشاء جدول داخل الداتابيز باسم 'customer_shopping_behavior'
    # إذا كان الجدول موجوداً مسبقاً، سيتم إضافة البيانات الجديدة إليه (append)
    table_name = 'customer_shopping_behavior'
    df.to_sql(name=table_name, con=engine, if_exists='append', index=False)
    
    print(f"🎉 مبروك! تم ربط البيانات ورفعها بنجاح إلى جدول ({table_name}) في SQL Server!")

except Exception as e:
    print("عذراً، حدث خطأ أثناء التنفيذ:")
    print(e)
```
